### Theoretical Background: Exact Enumeration for $N = 2$ Spins

To understand how the classical Ising Hamiltonian works computationally, we can exhaustively compute the energy of every possible state for a small system of **$N = 2$ spins**.

For a 2-spin system without an external field, the Hamiltonian simplifies to:
$$H(\sigma_1, \sigma_2) = -J \sigma_1 \sigma_2$$

* **State Space:** Each spin can take values $\sigma_i \in \{-1, +1\}$. For $N=2$, the total number of configurations is $2^2 = 4$, given by the Cartesian product $\{-1, +1\} \times \{-1, +1\}$.
* **Energy States (for $J > 0$):**
  * **Aligned States** ($(\uparrow, \uparrow)$ or $(\downarrow, \downarrow)$): $\sigma_1 \sigma_2 = +1 \implies E = -J$ (Lower energy / More stable state).
  * **Unaligned States** ($(\uparrow, \downarrow)$ or $(\downarrow, \uparrow)$): $\sigma_1 \sigma_2 = -1 \implies E = +J$ (Higher energy / Unstable state).

The code below generates all $2^2$ configurations using `itertools.product` and calculates their respective energies:

## Without Netket

```python
from itertools import product

J = 1.0 # Coupling constant

# Two sites, each with spin value -1 or +1
configurations = list(product([-1, 1], repeat=2))

for sigma1, sigma2 in configurations:
    energy = -J * sigma1 * sigma2
    print(f"Configuration: ({sigma1}, {sigma2}), Energy: {energy}")
```

### With Netket

```python
import netket as nk

# Two spin-1/2 sites
hi = nk.hilbert.Spin(s=1/2, N=2) # create a Hilbert space with 2 spin-1/2 sites
print(hi) # which system the operator belongs to 
# Define the Ising Hamiltonian
J = 1.0 

H = (
    -J 
    * nk.operator.spin.sigmaz(hi,0) # Spin operator for site 0 , z operator acting on site 0
    @ nk.operator.spin.sigmaz(hi,1) # Spin operator for site 1 , z operator acting on site 1
)

print("Number of sites:", hi.size)
print("Number of possible configurations:", hi.n_states)
print("Hamiltonian matrix:\n", H.to_dense())
```
